## 1. Data types (12 min)

A value has both a value and a **type**. The core types for this module are:

- `int`: whole number, e.g. number of trays `20`
- `float`: decimal number, e.g. flow rate `2.5`
- `str`: text, e.g. `'Reactor A'`
- `bool`: logical value, either `True` or `False`

Knowing the type helps prevent silent-looking mistakes.

```python
number_of_trays = 20
feed_flow_kg_h = 1250.5
unit_name = 'Absorber A'
pump_running = True

for value in [number_of_trays, feed_flow_kg_h, unit_name, pump_running]:
    print(value, '→', type(value))
```

```python
reactor_pressure = 5          
reactor_temperature = 185.5  
chemical_name = "Ethanol"     
valve_open = True            

for value in [reactor_pressure, reactor_temperature, chemical_name, valve_open]:
    print(value, "->", type(value))
```

```python
print(3 / 2, "->", type(3 / 2))
print(3 // 2, "->", type(3 // 2))
print(3 % 2, "->", type(3 % 2))
print(3 ** 2, "->", type(3 ** 2))
```

## 2. Arithmetic and comparison operators (15 min)

Arithmetic operators: `+`, `-`, `*`, `/`, `//` (whole-number division), `%` (remainder), `**` (power).

Comparison operators produce Boolean answers: `>`, `<`, `>=`, `<=`, `==` (is equal to), and `!=` (is not equal to). Note the difference between `=` for assignment and `==` for comparison.

```python
pressure_bar = 5.2
design_pressure_bar = 6.0

print('Pressure margin:', design_pressure_bar - pressure_bar, 'bar')
print('Above design pressure?', pressure_bar > design_pressure_bar)
print('Pressure exactly 5.2 bar?', pressure_bar == 5.2)
```

```python
feed = 800
product = 680

mass_yield = (product / feed) * 100

print("Mass yield:", mass_yield, "%")
print("At least 85%?", mass_yield >= 85)
```

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

## 3. Turning values into text and back (10 min)

The characters `25` can represent text or the number 25. Convert deliberately so Python knows which one you intend.

- `float('25.4')` converts numeric text to a decimal number.
- `int('25')` converts whole-number text.
- `str(25.4)` converts a number to text for a message.

An invalid conversion produces an error—useful information that the input needs checking.

```python
temperature_text = '72.5'
temperature_c = float(temperature_text)
temperature_k = temperature_c + 273.15
print('Temperature =', temperature_k, 'K')
```

## 4. Input and readable output (15 min)

`input()` pauses the program and returns **text**. Convert it before a calculation. An **f-string** inserts values inside text and can control decimal places, e.g. `{value:.2f}`.

```python
operator_name = input('Enter your name: ')
flow_rate_m3_h = float(input('Enter flow rate (m³/h): '))
hours = float(input('Enter operating time (h): '))
volume_m3 = flow_rate_m3_h * hours

print(f'{operator_name}, processed volume = {volume_m3:.2f} m³')
```

```python
mass_flow_rate = float(input("Enter mass flow rate (kg/h): "))
Cp = float(input("Enter specific heat capacity (kJ/kg.K): "))
delta_T = float(input("Enter temperature change (K): "))

heat_duty = mass_flow_rate * Cp * delta_T

print(f"Heat duty = {heat_duty:.2f} kJ/h")
```
